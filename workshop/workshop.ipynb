# MuJoCo

MuJoCo stands for Multi-Joint dynamics with Contact. It is a general purpose physics engine that aims to facilitate research and development in robotics, biomechanics, graphics and animation, machine learning, and other areas that demand fast and accurate simulation of articulated structures interacting with their environment. Initially developed by Roboti LLC, it was acquired and made [freely available](https://github.com/google-deepmind/mujoco/blob/main/LICENSE) by DeepMind in October 2021, and open sourced in May 2022. The MuJoCo codebase is available at the [google-deepmind/mujoco](https://github.com/google-deepmind/mujoco) repository on GitHub.

Key features:
1) Generalized coordinates combined with modern contact dynamics
2) Soft, convex and analytically-invertible contact dynamics
3) Tendon geometry
4) General actuation model
5) Reconfigurable computation pipeline
6) Model compilation
7) Separation of model and data
8) Interactive simulation and visualization
9) Powerful yet intuitive modeling language
10) Automated generation of composite flexible objects

https://mujoco.readthedocs.io/en/stable/overview.html

### XML: URDF and MJCF

**Extensible Markup Language (XML)** is a markup language and file format for storing, transmitting, and reconstructing arbitrary data.

The user defines models in the native **MJCF** scene description language – an XML file format designed to be as human readable and editable as possible. More detailed [description](https://mujoco.readthedocs.io/en/stable/XMLreference.html).

**Unified Robot Description Format (URDF)** is an XML format for representing a robot model. URDF is commonly used in Robot Operating System (ROS) tools such as rviz (Ros Visualization tool) and Gazebo simulator. The model consists of links and joints motion. More detailed [description](https://wiki.ros.org/urdf/XML/model).

To start `GUI` run the following command in terminal:

`python3 -m mujoco.viewer`

You can "drag-and-drop" model file in GUI or set path to model as parameter for command:

`python3 -m mujoco.viewer --mjcf=**filename**.xml`

You can load model directly from string that is described in `.py` file by using `mujoco.MjModel.from_xml_string()`

````python
xml_string = """
<mujoco>
  <option gravity ="0 0 -9.81" timestep="0.00001"/>
  <worldbody>
    <light name="top" pos="0 0 1"/>
    <body name="box_and_sphere" euler="0 0 -30">
      <joint name="swing" type="hinge" axis="1 -1 0" pos="-.2 -.2 -.2"/>
      <geom name="red_box" type="box" size=".2 .2 .2" rgba="1 0 0 1"/>
      <geom name="green_sphere" pos=".2 .2 .2" size=".1" rgba="0 1 0 1"/>
    </body>
  </worldbody>
</mujoco>
"""
model = mujoco.MjModel.from_xml_string(xml_string)
data = mujoco.MjData(model)
with mujoco.viewer.launch_passive(model, data) as viewer:
    while viewer.is_running():
        mujoco.mj_step(model, data)
        viewer.sync()
````

### Briefly about dynamics and integrators

MuJoCo computes forward and inverse dynamics in continuous time. 

Forward dynamics:
```math
\dot{v} = FD(q,v,\tau, f)
```

Inverse dynamics:
```math
(\tau, f^{ext}) = ID(q,\dot{q},\ddot{q})
```
More informaton about [forward](https://scaron.info/robotics/forward-dynamics.html) and [inverse](https://scaron.info/robotics/recursive-newton-euler-algorithm.html) dynamics.

The end result of forward dynamics is the joint acceleration $a = \dot{v}$. This is used to advance the simulation time from $t$ to $t+h$ and to update the state variables $q, v$. Four numerical integrators are available, three single-step integrators (`implicit Euler`, `implicitfast`, `implicit`) and the multi-step 4th order Runge-Kutta integrator (`RK4`). 

The explicit Euler method is not supported by MuJoCo but has pedagogical value. It can be written as:
```math
\text{velocity: } v_{t+h} = v_t + ha_t \\
\text{position: } q_{t+h} = q_t + hv_t
```

The accuracy and stability of all integrators can be improved by reducing the time step $h$. It is the most important parameter that the user can adjust. If it is too large, the simulation will become unstable. If it is too small, CPU time will be wasted without meaningful improvement in accuracy. 


Also, you can implement other integrator by calling `mj_forward` and integrating accelerations yourself.

See [Numerical Integration](https://mujoco.readthedocs.io/en/stable/computation/index.html#numerical-integration)


### Change model parameters

Model parameters can be changed directly (by manual) in `.xml` MJCF file. Another way is use [lxml](https://lxml.de/parsing.html) library inside that allows to parse .xml file and modify it. 

Also, MuJoCo has own struct for changing and creating models [mj_spec](https://mujoco.readthedocs.io/en/latest/programming/modeledit.html). But at now `mj_spec` has unstable API and we will not consider it here.

### Example 1: Change initial position of the red brick

````python
xml_tree = etree.parse('ex1.xml')
mjcf_container = xml_tree.getroot()
# Find body named "red_brick"
red_brick_container = mjcf_container.find(".//body[@name=\"red_brick\"]")

# Get initial offset of red brick
init_pos = red_brick_container.attrib['pos']
init_pos = init_pos.split() # separate string elements 
init_pos_num = map(float, init_pos) # convert from string to float type

# Set desired offset 
des_offset_num = np.array([0.5, 0, 0])
des_pos_num = list(init_pos_num)+des_offset_num
des_pos = np.array2string(des_pos_num)[1:-1] #convert to string without brackets

# Set desired pos to attribute pos for float_body
red_brick_container.attrib['pos'] = des_pos

with open(str('ex1_modify.xml'), "wb") as file:
    file.write(etree.tostring(xml_tree))
````

### Example 2: Change link L1 of manipulator and add new passive link L4

````python
xml_tree = etree.parse('ex5.xml')
mjcf_container = xml_tree.getroot()
# Find geom named "l1"
l1_container = mjcf_container.find(".//geom[@name=\"l1\"]")
# Set new parameters for l1
l1_new = 0.15
l1_color = "1 0 0 1"
l1_container.attrib['fromto'] = f'0 0 0 {l1_new} 0 0'
l1_container.attrib['rgba'] = l1_color

# If we change l1 lenght, we have to change joint q2 position by moving l2 body
l2_container = mjcf_container.find(".//body[@name=\"l2\"]")
l2_container.attrib['pos'] = f'{l1_new} 0 0'

# Find l3 body
l3_container = mjcf_container.find(".//body[@name=\"l3\"]")
# Add l4 body as subelement of l3
l4 = etree.SubElement(l3_container, "body")
# Set parameters of l4 body
l4.attrib['name'] = 'l4'
l4.attrib['pos'] = '.1 0 0'
# Add joint q4
j4 = etree.SubElement(l4, "joint")
j4.attrib['name'] = 'q4'
j4.attrib['damping'] = '0.005'
# Add geom parameters for l4
g4_1 = etree.SubElement(l4, "geom")
g4_1.attrib["name"] = 'g4'
g4_1.attrib["class"] = 'visual_cylinder'
g4_2 = etree.SubElement(l4, "geom")
g4_2.attrib["name"] = 'l4'
g4_2.attrib["fromto"] = '0 0 0 .1 0 0'
# Save changes to new file
with open(str('ex5_modify.xml'), "wb") as file:
    file.write(etree.tostring(xml_tree))
````
